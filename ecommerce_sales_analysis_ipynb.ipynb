```python
monthly_profit = df.groupby("month")["Profit"].sum().reset_index()

# Find Highest Profit Month
highest_profit = monthly_profit.loc[monthly_profit["Profit"].idxmax()]

print("Monthly Profit:")
print(monthly_profit)

print("\nHighest Profit Month:")
print(f"Month: {highest_profit['month']}")
print(f"Profit: ${highest_profit['Profit']:,.2f}")

# Plot Monthly Profit
plt.figure(figsize=(12,6))
bars = plt.bar(monthly_profit["month"].astype(str), monthly_profit["Profit"])

# Highlight Highest Profit Month
# The .name attribute would be the index of the highest profit row, not necessarily aligned with x-axis labels if month is a Period type.
# A more robust way to highlight is to iterate or find the index based on the month value.
# Let's find the index of the highest profit month in the plotted bars.
highest_profit_month_str = str(highest_profit['month'])
for i, bar in enumerate(bars):
    if monthly_profit["month"].astype(str).iloc[i] == highest_profit_month_str:
        bar.set_color("green")
        break

# Add value labels
for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        f"{bar.get_height():,.0f}",
        ha="center",
        va="bottom",
        fontsize=8,
        rotation=90
    )

plt.title("Monthly Profit Analysis")
plt.xlabel("Month")
plt.ylabel("Profit")
plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()

plt.show()
```

```python
category_profit = df.groupby("Category")["Profit"].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))
bars = plt.bar(category_profit.index, category_profit.values)

# Highlight Highest & Lowest
bars[0].set_color("green")   # Highest Profit
bars[-1].set_color("red")    # Lowest Profit

# Add Value Labels
for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        f"{bar.get_height():,.0f}",
        ha='center',
        va='bottom',
        fontsize=10
    )

plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

# -------------------------------
# Profit by Sub-Category
# -------------------------------
subcategory_profit = df.groupby("Sub-Category")["Profit"].sum().sort_values(ascending=False)

plt.figure(figsize=(12,6))
bars = plt.bar(subcategory_profit.index, subcategory_profit.values)

# Highlight Highest & Lowest
bars[0].set_color("green")   # Highest Profit
bars[-1].set_color("red")    # Lowest Profit

# Add Value Labels
for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        f"{bar.get_height():,.0f}",
        ha='center',
        va='bottom',
        fontsize=8,
        rotation=90
    )

plt.xticks(rotation=45, ha='right')
plt.title("Profit by Sub-Category")
plt.xlabel("Sub-Category")
plt.ylabel("Total Profit")
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

# -------------------------------
# Print Highest & Lowest Profit
# -------------------------------
print("Category with Highest Profit:")
print(category_profit.idxmax(), ":", round(category_profit.max(), 2))

print("\nCategory with Lowest Profit:")
print(category_profit.idxmin(), ":", round(category_profit.min(), 2))

print("\nSub-Category with Highest Profit:")
print(subcategory_profit.idxmax(), ":", round(subcategory_profit.max(), 2))

print("\nSub-Category with Lowest Profit:")
print(subcategory_profit.idxmin(), ":", round(subcategory_profit.min(), 2))
```

```python
segment_data = df.groupby("Segment")[["Sales", "Profit"]].sum().reset_index()

print(segment_data)

# Plot
x = np.arange(len(segment_data["Segment"]))
width = 0.35

plt.figure(figsize=(8,6))

plt.bar(x - width/2, segment_data["Sales"], width, label="Sales")
plt.bar(x + width/2, segment_data["Profit"], width, label="Profit")

# Value Labels
for i, v in enumerate(segment_data["Sales"]):
    plt.text(i - width/2, v, f"{v:,.0f}", ha="center", fontsize=9)

for i, v in enumerate(segment_data["Profit"]):
    plt.text(i + width/2, v, f"{v:,.0f}", ha="center", fontsize=9)

plt.xticks(x, segment_data["Segment"])
plt.title("Sales and Profit by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Amount ($)")
plt.legend()
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()

# Highest Sales Segment
highest_sales = segment_data.loc[segment_data["Sales"].idxmax()]
print("\nHighest Sales Segment:")
print(f"{highest_sales['Segment']} - ${highest_sales['Sales']:,.2f}")

# Highest Profit Segment
highest_profit = segment_data.loc[segment_data["Profit"].idxmax()]
print("\nHighest Profit Segment:")
print(f"{highest_profit['Segment']} - ${highest_profit['Profit']:,.2f}")
```

```python
total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()

sales_profit_ratio = total_sales / total_profit

print(f'Total Sales : ${total_sales:,.2f}')
print(f'Total Profit: ${total_profit:,.2f}')
print(f'Sales to Profit Ratio: {sales_profit_ratio:.2f}')

# -------------------------------
# Ratio by Customer Segment
# -------------------------------
segment_ratio = (
    df.groupby('Segment')[['Sales', 'Profit']]
      .sum()
      .reset_index()
)

# Calculate ratio
segment_ratio['Sales_to_Profit_Ratio'] = (
    segment_ratio['Sales'] / segment_ratio['Profit']
)

print('\nSales to Profit Ratio by Segment:')
print(segment_ratio)

# -------------------------------
# Graph
# -------------------------------
plt.figure(figsize=(8,5))
bars = plt.bar(segment_ratio['Segment'],
               segment_ratio['Sales_to_Profit_Ratio'])

# Highlight best (lowest ratio)
best_idx = segment_ratio['Sales_to_Profit_Ratio'].idxmin()
bars[best_idx].set_color('green')

# Add labels
for bar in bars:
    plt.text(bar.get_x() + bar.get_width()/2,
             bar.get_height(),
             f'{bar.get_height():.1f}',
             ha='center',
             va='bottom',
             fontsize=10)

plt.title('Sales to Profit Ratio by Customer Segment')
plt.xlabel('Customer Segment')
plt.ylabel('Sales / Profit Ratio')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

# Best segment
best_segment = segment_ratio.loc[best_idx]
print('\nBest Segment (Lowest Ratio):')
print(f"{best_segment['Segment']} → {best_segment['Sales_to_Profit_Ratio']:.2f}")
```
